```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
#Loading the data
df = pd.read_csv('spam.csv')
```

```python
df
```

```python
# 1. Ensure the column is of string type and then clean the strings: strip white spaces and convert to lowercase
df['label'] = df['label'].astype(str).str.strip().str.lower()

# 2. Apply the map function
df['label'] = df['label'].map({'spam': 1, 'ham': 0})
```

```python
#There were duplicate data
df = df.drop_duplicates()
```

```python
df
```

```python
#Splitting the data in labels and features
X = df['message']
Y = df['label']
```

```python
#Creating train test split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42, stratify=Y)
```

```python
#Importing TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

```python
#Logistic Regression
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(class_weight='balanced')
model.fit(tfidf_matrix,Y_train)
predictions = model.predict(X_test_tfidf)
```

```python
#Evaluation
from sklearn.metrics import recall_score
recall_score(Y_test,predictions)
```

```python
#importing confusion matrix
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Y_test, predictions)
print(cm)
```
